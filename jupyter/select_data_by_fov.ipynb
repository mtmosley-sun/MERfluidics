```python
from shutil import copy2
import re
import os, sys
```

```python
kept_fovs = list(range(0,3)) + [6,7] +\
list(range(15,22)) + \
list(range(28,31)) + \
[40,41] + \
list(range(48,56)) + \
list(range(80,94)) + \
list(range(115,121)) + \
list(range(187,194)) + \
list(range(201,221))
```

```python
len(kept_fovs)
```

```python
source_folder = r'D:\Data\Pu\20260215-HEK-LP_TMD1large_TMD1-TMDER1-multiIF_tritonX\20260218_110248_550'
target_folder = r'D:\Data\Pu\20250306-4T1FE_MF10\20260219'
if not os.path.exists(target_folder):
    print(f'Creating target folder: {target_folder}')
    os.makedirs(target_folder)
```

```python
576/16
```

```python
re_string = r"(Conv|Confocal)_Hyb([0-9]+)_([0-9]+).nd2"
copy_file_count = 0

for _fl in os.listdir(source_folder):
    if re.match(re_string, _fl):
        fov = int(re.match(re_string, _fl).group(3))
        if fov in kept_fovs:
            target_fov = kept_fovs.index(fov)
            print(f"source: {_fl}, target_file:{re.sub(re_string, r'\1_Hyb\2_' + str(target_fov) + '.nd2', _fl)}")
            copy2(os.path.join(source_folder, _fl), 
                  os.path.join(target_folder, re.sub(re_string, r'\1_Hyb\2_' + str(target_fov) + '.nd2', _fl)))
            copy_file_count += 1
    #break
print(f"Total {copy_file_count} files copied.")
```
